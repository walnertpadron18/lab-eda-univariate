```python
price = df["price"]

mean_price = round(price.mean(), 2)
median_price = round(price.median(), 2)
mode_price = round(price.mode()[0], 2)

print(f"Media: £{mean_price}")
print(f"Mediana: £{median_price}")
print(f"Moda: £{mode_price}")
```

La media de precio por productos es de 89 y la moda de 9.99. hay un sesgo muy grande hacia la derecha por en el precio

```python
variance = price.var()
std_dev = price.std()


price_range = price.max() - price.min()


q1 = price.quantile(0.25)
q3 = price.quantile(0.75)
iqr = q3 - q1

print(f"Varianza: {variance:.2f}")
print(f"Desviación estándar: £{std_dev:.2f}")
print(f"Rango: £{price_range:.2f} (mín £{price.min():.2f}, máx £{price.max():.2f})")
print(f"Q1: £{q1:.2f} | Q3: £{q3:.2f} | IQR: £{iqr:.2f}")
```

Existe una dispersion significativa en los precios
